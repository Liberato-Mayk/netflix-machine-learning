```python
import pandas as pd
df = pd.read_csv("netflix.csv", sep=None, engine="python")
```

```python
df["release_year"] = df["release_year"].astype("Int64")
```

```python
# Eliminar caracteres BOM y espacios de los nombres de columnas
df.columns = df.columns.str.replace('\ufeff', '', regex=False).str.strip()
```

```python
df = df[["show_id", "type", "title", "date_added", "release_year", "rating", "duration"]]
```

```python
# convertimos a formato fecha
df["date_added"] = pd.to_datetime(df["date_added"], dayfirst=True, errors="coerce")
```

```python
df
```

```python
# extraemos el año de la fecha de agregado
df["year_added"] = df["date_added"].dt.year.astype("Int64")
```

```python
df
```

```python
# extraemos el día de la fecha de agregado
df["day_added"] = df["date_added"].dt.day.astype("Int64")
```

```python
df.head()
```

```python
df["years_to_netflix"] = (df["year_added"] - df["release_year"]).astype("Int64")
```

```python
df
```

```python
# reemplazamos los valores faltantes de la columna rating por "Desconocido"
df["rating"] = df["rating"].fillna("Desconocido")
```

```python
df["month_added"] = df["date_added"].dt.month.astype("Int64")
```
